## Straight from a plotting function

No files needed: every combination is rendered once, up front.

```python
import numpy as np
import matplotlib.pyplot as plt

def plot(freq, phase, noise):
    x = np.linspace(0, 4 * np.pi, 400)
    rng = np.random.default_rng(0)
    fig, ax = plt.subplots(figsize=(6, 3.5))
    ax.plot(x, np.sin(freq * x + phase) + noise * rng.standard_normal(x.size))
    ax.set_ylim(-2, 2)
    ax.set_title(f'freq={freq}  phase={phase:.2f}  noise={noise}')
    return fig

Viewer.from_function(plot, freq=[1, 2, 3], phase=[0, np.pi / 4, np.pi / 2], noise=[0, 0.1, 0.3])
```
